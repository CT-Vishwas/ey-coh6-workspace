When we read a data from a data source, we often want to perform some operations on it before using it in our application. This process is called data loading and preprocessing. In this article, we will discuss how to load and preprocess data in Python using the `pandas` library.

Preprocessing step include:
- Handling missing values
- Handling outliers
- Encoding categorical variables
- Feature scaling
- Feature engineering
- Data transformation
- Data splitting

```python
import pandas as pd
```

```python
# dataframe with missing values, wrong formats, and outliers. We will use the `pandas` library to perform these operations.
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve'],
    'Age': [25, 30, None, 35, 40],
    'Salary': ["50000", "60000", "55000", None, "70000"],
    'Department': ['HR', 'Engineering', 'Marketing', 'Engineering', 'HR']
}
```

```python
df = pd.DataFrame(data)
# Display the original dataframe
print("Original DataFrame:")
print(df)
```

```python
df.info()
```

To Fix missing values:
- We can use the `fillna()` method to fill missing values with a specific value or with the mean, median, or mode of the column.
- We can use the `dropna()` method to drop rows or columns with missing values.

```python
df1 = df.copy()
# To Fix missing values:
df1.dropna(inplace=True)
df1.head()
```

```python
df2 = df.copy()
df2.fillna(value={"Age":18, "Salary":0}, inplace=True)
df2.head()
```

```python
df2.info()
```

```python
df2['Salary'] = df2['Salary'].astype('Float64')
df2.info()
```

```python
import matplotlib.pyplot as plt
```

```python
bar_colors = ['tab:red', 'tab:blue', 'tab:red', 'tab:orange', 'tab:red']
df2.plot(kind='bar',x="Name",y="Salary", rot=45, color='blue')
plt.xlabel('Employee Names')
plt.ylabel("Employee Salary")
plt.title("Salary Trends")
# plt.grid(color='cyan')
plt.show()
plt.savefig("../outputs/salary_trends.png")
```
